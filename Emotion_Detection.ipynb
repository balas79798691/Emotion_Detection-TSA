# Text Emotion Detection Application

An application that analyzes text and identifies its dominant emotional category.

## 1. Libraries

```python
import re
import string
import pandas as pd
import matplotlib.pyplot as plt

print("Emotion Detection application is ready.")
```

## 2. Emotion Knowledge Base

```python
emotion_words = {
    "Joy": {"happy","joy","excited","wonderful","amazing","love","great","excellent","glad","success","win","celebrate"},
    "Sadness": {"sad","cry","crying","unhappy","lonely","hurt","loss","miss","heartbroken","disappointed","grief","upset"},
    "Anger": {"angry","anger","hate","furious","annoyed","rage","irritated","mad","frustrated","offended"},
    "Fear": {"afraid","fear","scared","terrified","danger","worried","anxious","nervous","panic","threat"},
    "Surprise": {"surprised","unexpected","shocked","astonished","suddenly","wow","unbelievable","amazed"},
    "Love": {"love","loving","romantic","care","caring","affection","dear","passion","kindness","relationship"}
}

print("Categories:", ", ".join(emotion_words))
```

## 3. Emotion Detection Engine

```python
def preprocess(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = text.translate(str.maketrans("", "", string.punctuation))
    return text.split()

def emotion_scores(text):
    words = preprocess(text)
    return {
        emotion: sum(word in vocabulary for word in words)
        for emotion, vocabulary in emotion_words.items()
    }

def detect_emotion(text):
    scores = emotion_scores(text)
    highest = max(scores.values())

    if highest == 0:
        return "Neutral", scores

    return max(scores, key=scores.get), scores

sample = "I am extremely happy and excited about the wonderful success of our project."
emotion, scores = detect_emotion(sample)
print("Detected:", emotion)
print("Scores:", scores)
```

## 4. Application

```python
def emotion_analyzer():
    print("=" * 60)
    print("                 EMOTION ANALYZER")
    print("=" * 60)
    print("Type 'quit' to exit.\n")

    while True:
        text = input("Enter your text: ")

        if text.lower() == "quit":
            print("Application closed.")
            break

        if not text.strip():
            print("Please enter some text.")
            continue

        emotion, scores = detect_emotion(text)

        print("\n--- RESULT ---")
        print("Dominant Emotion:", emotion)

        for name, score in sorted(scores.items(), key=lambda x: x[1], reverse=True):
            print(f"{name:10s}: {score}")

        print("-" * 60)

# Uncomment to run:
# emotion_analyzer()
```

## 5. Test Cases

```python
sentences = [
    "I am extremely happy because I won the competition.",
    "I feel lonely and sad after losing my favorite book.",
    "I am furious because the service was terrible.",
    "I am nervous about tomorrow's examination.",
    "Wow! I never expected this amazing surprise.",
    "I deeply care about my family and friends.",
    "The meeting starts at ten o'clock."
]

records = []
for text in sentences:
    emotion, scores = detect_emotion(text)
    records.append({
        "Text": text,
        "Emotion": emotion,
        "Score": max(scores.values())
    })

test_df = pd.DataFrame(records)
display(test_df)

plt.figure(figsize=(8,4))
test_df["Emotion"].value_counts().plot(kind="bar")
plt.title("Emotion Distribution")
plt.xlabel("Emotion")
plt.ylabel("Number of Samples")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Conclusion

The application identifies the dominant emotion in text using an interpretable emotion lexicon. It can later be extended using a machine-learning model and a larger labeled dataset.
